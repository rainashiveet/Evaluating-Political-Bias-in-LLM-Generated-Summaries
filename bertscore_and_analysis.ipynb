```python
from bert_score import score
import pandas as pd

models = [
    "gpt4o_abstractive_summary",
    "gemini_abstractive_summary",
    "llama_abstractive_summary",
    "deepseek_abstractive_summary",
    "mistral_abstractive_summary",
    "claude_abstractive_summary"
]

reference = df["bert_extractive_summary"].astype(str).tolist()

bert_results = {}

for model in models:
    
    prediction = df[model].astype(str).tolist()
    
    P, R, F1 = score(prediction, reference, lang="en")
    
    bert_results[model] = F1.mean().item()

bert_df = pd.DataFrame.from_dict(
    bert_results,
    orient="index",
    columns=["BERTScore_F1"]
)

bert_df
```

```python
import matplotlib.pyplot as plt

bert_df.plot(kind="bar", figsize=(8,5))

plt.title("BERTScore Comparison Across LLMs")
plt.ylabel("BERTScore (F1)")
plt.xlabel("Models")

plt.xticks(rotation=45)

plt.show()
```

# FAIRNESS ANALYSIS (Political Bias)

```python
entities = ["modi", "rahul"]
```

# Sentiment function (VADER)

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

def entity_sentiment(text, entity):
    
    text = str(text).lower()
    
    if entity in text:
        return analyzer.polarity_scores(text)["compound"]
    
    return 0
```

# Compute fairness

```python
fairness_results = {}

for model in models:
    
    modi_scores = []
    rahul_scores = []
    
    for text in df[model]:
        
        modi_scores.append(entity_sentiment(text, "modi"))
        rahul_scores.append(entity_sentiment(text, "rahul"))
    
    fairness_results[model] = {
        "Modi Sentiment": sum(modi_scores)/len(modi_scores),
        "Rahul Sentiment": sum(rahul_scores)/len(rahul_scores)
    }

fairness_df = pd.DataFrame(fairness_results).T

fairness_df
```

# FAIRNESS GRAPH

```python
fairness_df.plot(kind="bar", figsize=(8,5))

plt.title("Political Entity Sentiment Comparison")
plt.ylabel("Sentiment Score")
plt.xlabel("Models")

plt.xticks(rotation=45)

plt.show()
```
